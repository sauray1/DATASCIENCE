```python
import numpy as np
import time
np.random.seed(42)
print("NumPy version:",np.version.version)
```

```python
prices =list(range(1,1000001))

start =time.time()
discounted_loop=[p*0.95 for p in prices]
loop_time=time.time() - start
print(f"loop version:{loop_time:.4f} seconds")

prices_np =np.arange(1,1000001)
discounted_loop=[prices_np*0.95]
np_time=time.time() - start
print(f"Numpy version:{np_time:.4f} seconds")
print(f"Numpy was about{loop_time/np_time:.0f}%faster")
```
